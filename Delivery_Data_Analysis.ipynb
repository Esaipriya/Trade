My goal is to analyze delivery time and order patterns

Focus Only on the Columns  Need for three analysis questions,
 only need these columns:
Created At, Delivered At, City,Customer, Order ID

```python
df[['Created At', 'Delivered At', 'City', 'Customer', 'Order ID']].isnull().sum()
```

```python
# Convert date columns
df['Created At'] = pd.to_datetime(df['Created At'], errors='coerce')
df['Delivered At'] = pd.to_datetime(df['Delivered At'], errors='coerce')
df = df.dropna(subset=['Created At', 'Delivered At'])
```

```python
# Calculate delivery days
df['Delivery_Days'] = (df['Delivered At'] - df['Created At']).dt.total_seconds() / (24 * 3600)
```

```python
df['Delivery_Days']
```

```python
# Remove negative values
df = df[df['Delivery_Days'] >= 0]
```

```python
# Check result
df[['City', 'Created At', 'Delivered At', 'Delivery_Days']].head()
```

Q1: Average Delivery Time by City

```python
avg_delivery_by_city = df.groupby('City')['Delivery_Days'].mean().reset_index().sort_values('Delivery_Days')
avg_delivery_by_city.head(10)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.barplot(x='City', y='Delivery_Days', data=avg_delivery_by_city)
plt.xticks(rotation=90)
plt.title("Average Delivery Time by City")
plt.xlabel("City")
plt.ylabel("Average Delivery Days")
plt.show()
```

Q2: Late Deliveries by City

```python
# Merge average delivery times
df = df.merge(avg_delivery_by_city, on='City', suffixes=('', '_Avg'))

# Mark late deliveries
df['Is_Late'] = (df['Delivery_Days'] > df['Delivery_Days_Avg']).astype(int)

# Count late deliveries
late_by_city = df.groupby('City')['Is_Late'].sum().reset_index().sort_values('Is_Late', ascending=False)
late_by_city.head(10)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='City', y='Is_Late', data=late_by_city)
plt.xticks(rotation=90)
plt.title("Late Deliveries by City")
plt.ylabel("Count of Late Deliveries")
plt.xlabel("City")
plt.show()
```

Q3: Customer Order Frequency

```python
# Count orders per customer
customer_orders = df.groupby('Customer')['Order ID'].nunique().reset_index(name='Order_Count')

# Categorize customers
def categorize(count):
    if count == 1:
        return 'Once'
    elif count == 2:
        return 'Twice'
    elif count == 3:
        return 'Thrice'
    else:
        return 'More than 4 times'

customer_orders['Category'] = customer_orders['Order_Count'].apply(categorize)

# Count categories
order_frequency = customer_orders['Category'].value_counts().reset_index()
order_frequency.columns = ['Order_Frequency', 'Customer_Count']
order_frequency
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='Order_Frequency', y='Customer_Count', data=order_frequency, palette='Set2')
plt.title("Customer Order Frequency")
plt.ylabel("No. of Customers")
plt.xlabel("Order Frequency")
plt.show()
```

## Key Insights

- The average delivery time varies significantly by city — certain regions show faster logistics.
- Cities with higher late deliveries may require route optimization.
- Most customers ordered only once, indicating opportunities for customer retention campaigns.
